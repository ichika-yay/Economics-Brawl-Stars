**奔騰身為一個派對機器人，需要攝取能量飲料和機油維持運轉。麥克絲和梅格負責去雷霆小鎮購買能量飲給奔騰，假設機油為正常財。**

**彈性模擬器:** 假設能量飲需求函數為線性函數 P=a-bQ,讓使用者自行輸入a,b值,導出需求函數<br>
1.用滑桿選取兩點，計算弧彈性<br>
2.用滑桿選取一點，計算點彈性


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact
from google.colab import output
output.enable_custom_widget_manager()
def ci(prompt):
  while True:
    try:
      value=float(input(prompt))
      if value <= 0:
        print("請輸入正數,否則不符合需求法則")
        continue
      return value
    except ValueError:
      print("輸入錯誤 請輸入有效數值")
print("假設能量飲為線性需求函數:P=a-bQ,請輸入:")
a=ci("a(價格軸的截距)為多少?")
b=ci("b(需求線斜率絕對值)為多少?")
def equillibrium(Q0=(a/b)*0.7,Q1=(a/b)/4):
  P0=a-b*Q0
  P1=a-b*Q1
  plt.scatter([Q0,Q1,(a/b)/2],[P0,P1,a/2],color=["red","green","blue"],s=80,zorder=5)
  plt.annotate("E0",(Q0,P0),textcoords="offset points",xytext=(10,10))
  plt.annotate("E1",(Q1,P1),textcoords="offset points",xytext=(10,10))
  x_axis=np.linspace(0.1,(a/b)*1.1,300)
  y_axis=a-b*x_axis
  plt.plot(x_axis,y_axis,"b-",label="Demand Curve")
  plt.xlim(0,(a/b)*1.1)
  plt.ylim(0,a*1.1)
  plt.xlabel("Quantity")
  plt.ylabel("Price")
  plt.title("Energy Drink Demand Curve")
  plt.legend()
  plt.grid(True,alpha=0.3)
  def elastic(Q,P):
    try:
      return (1/b)*(P/Q)
    except ZeroDivisionError:
      return "infinite"
  Ed_0=elastic(Q0,P0)
  Ed_1=elastic(Q1,P1)
  Ed_arc=(1/b)*((P0+P1)/(Q0+Q1))
  print("需求價格彈性如下:")
  if Q0==Q1:
    print("兩點交合，弧彈性不存在")
  else:
    print(f"E0&E1兩點的弧彈性為{Ed_arc:.3f}")
  def Eprint(elas,point):
    if elas=="infinite":
      print(f"點{point}的點彈性為無限大")
    else:
      print(f"點{point}的點彈性為{elas:.3f}")
      print(f"在這個瞬間，價格如果漲價1%，需求量將會下降{elas:.3f}%")
      if elas>1:
        print("該點在此刻富有彈性")
      elif elas==1:
        print("該點在此刻呈單位彈性")
      elif elas==0:
        print("該點在此刻完全無彈性")
      else:
        print("該點在此刻缺乏彈性")
  Eprint(Ed_0,"E0")
  Eprint(Ed_1,"E1")
  plt.tight_layout()
  plt.show()
  plt.close()
interact(
    equillibrium,
    Q0=widgets.FloatSlider(value=(a/b)*0.7,min=0,max=a/b,description="Old Quantity"),
    Q1=widgets.FloatSlider(value=(a/b)/4,min=0,max=a/b,description="New Quantity"),
);
```

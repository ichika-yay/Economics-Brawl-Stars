**奔騰身為一個派對機器人，需要攝取能量飲料和機油維持運轉。
麥克絲和梅格負責去雷霆小鎮購買能量飲給奔騰，假設機油為正常財。**

**希克斯分解:** 讓使用者自行用滑桿控制偏好權重,Y(機油)價格, X(能量飲)價格變化, 所得等<br>列出替代效果&所得效果並在圖表上可視化, 導出需求曲線並標出對應點

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact
from google.colab import output
output.enable_custom_widget_manager()
# 引進動態互動元件，經濟參數能用滑桿控制&Colab支援模組
def formula(alpha=0.5,beta=0.5,I=100,Py=5,Px0=10.0,Px1=4.0):
# 定義繪圖函式，幫經濟參數設定預設值
  ratio_a=(beta/alpha)*(Px0/Py)
# 利用MUx/MUy=Px/Py解出預算花光&效用最大化下 Y等於幾倍的X 並代Px0
  X_A=(I*alpha)/((alpha+beta)*Px0)
  Y_A=ratio_a*X_A
  U_0=(X_A**alpha)*(Y_A**beta)
# 代Px0到X的需求函數裡 導出舊最適消費量&效用值
  ratio_c=(beta/alpha)*(Px1/Py)
  X_C=(I*alpha)/((alpha+beta)*Px1)
  Y_C=ratio_c*X_C
  U_1=(X_C**alpha)*(Y_C**beta)
# 倍數函數和需求函數代Px1，導出新最適消費量&效用值
  X_B=(U_0/(ratio_c**beta))**(1.0/(alpha+beta))
  Y_B=ratio_c*X_B
# 替代效果Y&X倍數關係和新組合一致(因都是用Px1/Py計算),效用與舊無異曲線一致
  SE=X_B-X_A
  IE=X_C-X_B
  TE=X_C-X_A
  print("希克斯分解結果")
  print(f"初始點:能量飲{X_A:.2f}瓶,機油{Y_A:.2f}瓶,效用值{U_0:.2f}")
  print(f"補償點:能量飲{X_B:.2f}瓶,機油{Y_B:.2f}瓶")
  print(f"最終點:能量飲{X_C:.2f}瓶,機油{Y_C:.2f}瓶,效用值{U_1:.2f}")
  print(f"替代效果={SE:.2f}瓶能量飲")
  print(f"所得效果={IE:.2f}瓶能量飲")
  print(f"最終效果={TE:.2f}瓶能量飲")
  fig,(ax1,ax2)=plt.subplots(2,1,figsize=(8,9))
# subplot用來建立大圖裡的子圖(希克斯分解圖&需求曲線)
# plt.subplot(列,欄) 回傳值=畫布,子圖物件
  X_max=max(X_A,X_B,X_C)*1.5
  x_value=np.linspace(0.1,X_max,300)
# 設下限防止下式ZeroDivisionError Cobb-Douglas也不會到達X=0
  ax1.plot(x_value,(U_0/(x_value**alpha))**(1/beta),"g-",label=f"U0 {U_0:.2f}")
  ax1.plot(x_value,(U_1/(x_value**alpha))**(1/beta),"g-",label=f"U1 {U_1:.2f}")
# 繪製無異曲線
# 語法"r-","g--"等等為顏色+線條形式
  ax1.plot(x_value,(I-Px0*x_value)/Py,"m-",label="Old Budget Line")
  ax1.plot(x_value,(I-Px1*x_value)/Py,"b-",label="New Budget Line")
  I_B=Px1*X_B+Py*Y_B
  ax1.plot(x_value,(I_B-Px1*x_value)/Py,"b--",alpha=0.6)
# 繪製預算線 桃紅:舊 藍:替代&新預算線
  ax1.scatter([X_A,X_B,X_C],[Y_A,Y_B,Y_C],color=["red","blue","green"],s=80,zorder=5)
# 語法允許用中括號一起執行
  ax1.annotate("E0",(X_A,Y_A),textcoords="offset points",xytext=(10,10))
  ax1.annotate("SE",(X_B,Y_B),textcoords="offset points",xytext=(10,10))
  ax1.annotate("SE+IE(E1)",(X_C,Y_C),textcoords="offset points",xytext=(10,10))
# 標示出三個關鍵點
# annotate為幫點新增註解 語法:(文字,位置,形式(此為"文字偏離資料點",偏離程度(x,y))
  ax1.set_xlim(0,X_max)
  ax1.set_ylim(0,max(Y_A,Y_B,Y_C)*1.5)
  ax1.set_xlabel("Energy Drink")
  ax1.set_ylabel("Robot Oil")
  ax1.set_title("Hicksian Decomposition")
  ax1.legend()
  ax1.grid(True,alpha=0.3)
# set為物件ax的命名規則
  p_value=np.linspace(0.1,max(Px0,Px1)*1.5,300)
  x_demand=(I*alpha)/((alpha+beta)*p_value)
  ax2.plot(x_demand,p_value,color="purple",linewidth=2,label="Demand Curve")
  ax2.scatter([X_A,X_C],[Px0,Px1],color=["red","green"],s=80,zorder=5)
  ax2.annotate(f"{X_A:.2f},{Px0:.2f}",(X_A,Px0),textcoords="offset points",xytext=(10,10))
  ax2.annotate(f"{X_C:.2f},{Px1:.2f}",(X_C,Px1),textcoords="offset points",xytext=(10,10))
# 需求曲線 X軸數量 Y軸價格
  ax2.set_xlim(0,X_max)
  ax2.set_ylim(0,max(Px0,Px1)*1.5)
  ax2.set_xlabel("Energy Drink")
  ax2.set_ylabel("Energy Drink Price")
  ax2.set_title("Demand Curve")
  ax2.legend()
  ax2.grid(True,alpha=0.3)
  plt.tight_layout()
  plt.show()
  plt.close(fig)
# 讓程式在每次滑動滑桿後都能重新繪製圖表
interact(
    formula,
    alpha=widgets.FloatSlider(value=0.5,min=0.1,max=2.0,step=0.1,description="X Preference"),
    beta=widgets.FloatSlider(value=0.5,min=0.1,max=2.0,step=0.1,description="Y Preference"),
    I=widgets.FloatSlider(value=100,min=20,max=200,step=10,description="Income"),
    Py=widgets.FloatSlider(value=5,min=1,max=20,step=1,description="Robot Oil Price"),
    Px0=widgets.FloatSlider(value=10,min=1,max=20,step=1,description="Drink Old Price"),
    Px1=widgets.FloatSlider(value=4,min=1,max=20,step=1,description="Drink New Price"),
);
# 設定滑桿 語法:(初始值,最小值,最大值,移動刻度,文字)
```
